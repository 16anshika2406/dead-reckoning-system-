```python
import osmnx as ox
import networkx as nx
import matplotlib.pyplot as plt

# 1. Calculate boundaries from the dataset
west = df['GPS LONGITUDE (degrees)'].min() - 0.01
south = df['GPS LATITUDE (degrees)'].min() - 0.01
east = df['GPS LONGITUDE (degrees)'].max() + 0.01
north = df['GPS LATITUDE (degrees)'].max() + 0.01

# 2. Define bbox tuple in (left, bottom, right, top) order required by current OSMnx versions
bbox = (west, south, east, north)

# 3. Download road network graph
G = ox.graph_from_bbox(bbox, network_type='drive')
print("Road network downloaded successfully! Total nodes:", len(G.nodes))

# 4. Extract raw coordinates
lats = df['GPS LATITUDE (degrees)'].values
lons = df['GPS LONGITUDE (degrees)'].values

# 5. Snap GPS points to the nearest road network nodes
nearest_nodes = ox.distance.nearest_nodes(G, X=lons, Y=lats)

# 6. Extract matched coordinates
matched_lats = [G.nodes[node]['y'] for node in nearest_nodes]
matched_lons = [G.nodes[node]['x'] for node in nearest_nodes]

# 7. Plot original raw GPS versus map-matched road path
plt.figure(figsize=(8, 6))
plt.plot(lons, lats, color='red', alpha=0.5, label='Raw GPS Track')
plt.plot(matched_lons, matched_lats, color='blue', linewidth=2, label='Map-Matched Road Path')
plt.legend()
plt.title('Map-Matching: Raw GPS vs. Snapped Road Network')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import osmnx as ox
from shapely.geometry import Point, LineString
import numpy as np

# Load data
file_path = "IO-VNBD/Unsynchronised V and S Dataset/Uncategorised IOVNB (V and S) Dataset/S-Dataset/S-M.csv"
df = pd.read_csv(file_path, encoding="latin-1")

lat_col = [c for c in df.columns if "LAT" in c.upper()][0]
lon_col = [c for c in df.columns if "LON" in c.upper()][0]

# Road network sized to the trip
north = df[lat_col].max() + 0.01
south = df[lat_col].min() - 0.01
east = df[lon_col].max() + 0.01
west = df[lon_col].min() - 0.01
graph = ox.graph_from_bbox((west, south, east, north), network_type="drive")
print("Road network loaded:", len(graph.nodes), "nodes")

# VECTORIZED nearest-edge lookup — all points at once, much faster than a loop
lats = df[lat_col].values
lons = df[lon_col].values

nearest = ox.distance.nearest_edges(graph, X=lons, Y=lats)  # batch call
print("Nearest edges found for all", len(lats), "points")

# Now snap each point onto its matched road (this loop is fast, no network calls)
matched_coords = []
for i in range(len(lats)):
    u, v, key = nearest[i]
    edge = graph.edges[u, v, key]
    line = edge.get('geometry', LineString([(graph.nodes[u]['x'], graph.nodes[u]['y']),
                                             (graph.nodes[v]['x'], graph.nodes[v]['y'])]))
    pt = Point(lons[i], lats[i])
    snapped = line.interpolate(line.project(pt))
    matched_coords.append((snapped.y, snapped.x))

print(f"Matched all {len(matched_coords)} points")

# Plot
plt.figure(figsize=(10,8))
plt.plot(df[lon_col], df[lat_col], 'r-', linewidth=1, label='Raw GPS')
plt.plot([c[1] for c in matched_coords], [c[0] for c in matched_coords],
         'g-', linewidth=2, label='Map-Matched')
plt.legend()
plt.title("Map-Matching Result — Full Resolution")
plt.axis('equal')
plt.show()

# Save
matched_df = pd.DataFrame(matched_coords, columns=["matched_lat", "matched_lon"])
matched_df.to_csv("map_matching_output.csv", index=False)
print("Saved:", matched_df.shape)
```
